```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("Cleaned_APL_Logistics.csv")
```

FEATURE 1 – DELIVERY GAP
Delivery Gap =
Actual Shipping Days − Scheduled Shipping Days

```python
df["Delivery Gap"] = (
    df["Days for shipping (real)"]
    - df["Days for shipment (scheduled)"]
)
```

```python
df["Delivery Gap"].head()
```

FEATURE 2 – SHIPPING PRESSURE INDEX

```python
df["Shipping Pressure Index"] = (
    df["Order Item Quantity"] *
    df["Product Price"]
) / (df["Days for shipment (scheduled)"] + 1)
```

FEATURE 3 – DISCOUNT PERCENTAGE

```python
df["Discount Percentage"] = (
    df["Order Item Discount Rate"] * 100
)
```

FEATURE 4 – ORDER VALUE CATEGORY

```python
df["Order Value Category"] = pd.qcut(
    df["Sales"],
    q=3,
    labels=["Low","Medium","High"]
)
```

FEATURE 5 – PROFIT CATEGORY

```python
df["Profit Category"] = pd.qcut(
    df["Order Profit Per Order"],
    q=3,
    labels=["Low","Medium","High"]
)
```

FEATURE 6 – HIGH VALUE CUSTOMER FLAG

```python
median_sales = df["Sales per customer"].median()

df["High Value Customer"] = np.where(
    df["Sales per customer"] >= median_sales,
    1,
    0
)
```

FEATURE 7 – EXPRESS SHIPPING FLAG

```python
df["Express Shipping"] = np.where(
    df["Shipping Mode"].isin(["Same Day", "First Class"]),
    1,
    0
)
```

FEATURE 8 – LARGE ORDER FLAG

```python
df["Large Order"] = np.where(
    df["Order Item Quantity"] >= 5,
    1,
    0
)
```

FEATURE 9 – PROFIT MARGIN

```python
df["Profit Margin"] = np.where(
    df["Sales"] == 0,
    0,
    df["Order Profit Per Order"] / df["Sales"]
)
```

```python
df["Profit Margin"] = df["Profit Margin"].replace(
    [np.inf,-np.inf],
    0
)

df["Profit Margin"] = df["Profit Margin"].fillna(0)
```

FEATURE 10 – SALES CATEGORY

```python
df["Sales Category"] = pd.cut(
    df["Sales"],
    bins=[0,100,300,600,5000],
    labels=["Low","Medium","High","Premium"]
)
```

FEATURE 11 – CUSTOMER VALUE SCORE

```python
df["Customer Value Score"] = (
    df["Sales per customer"] *
    df["Order Profit Per Order"]
)
```

FEATURE 12 – SHIPPING EFFICIENCY

```python
df["Shipping Efficiency"] = np.where(
    df["Days for shipment (scheduled)"] == 0,
    1,
    df["Days for shipping (real)"] /
    df["Days for shipment (scheduled)"]
)
```

FEATURE 13 – ORDER COMPLEXITY SCORE

```python
df["Order Complexity Score"] = (
    df["Order Item Quantity"] *
    (1 + df["Order Item Discount Rate"]) *
    df["Product Price"]
)
```

FEATURE 14 – MARKET RISK SCORE

```python
market_risk = df.groupby("Market")["Late_delivery_risk"].mean()

df["Market Risk Score"] = df["Market"].map(market_risk)
```

FEATURE 15 – SHIPPING MODE RISK

```python
mode_risk = df.groupby("Shipping Mode")["Late_delivery_risk"].mean()

df["Shipping Mode Risk"] = df["Shipping Mode"].map(mode_risk)
```

FEATURE 16 - DELIVERY PERFORMANCE

```python
df["Delivery Performance"] = np.where(
    df["Delivery Gap"] <= 0,
    "On Time",
    "Delayed"
)
```
